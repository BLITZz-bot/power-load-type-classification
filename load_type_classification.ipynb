```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data/load_data.csv')

print("1. Shape of raw data (rows, columns):", df.shape)
print("\n2. Columns with Missing Values (Blanks):")
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
df
```

```python
df.columns = df.columns.str.strip()
```

```python
df['Load_Type'] = df['Load_Type'].astype(str).str.strip()
df
```

```python
df = df.replace(r'^\s*$', np.nan, regex=True).replace('?', np.nan)
df
```

```python
df = df.ffill().bfill()
```

```python
print(df.isnull().sum())
df
```

```python
# Convert Date_Time to datetime format
df['Date_Time'] = pd.to_datetime(df['Date_Time'], format='%d-%m-%Y %H:%M')

# 1. Hour of the day (0 to 23)
df['hour'] = df['Date_Time'].dt.hour

# 2. Day of the week (0 = Monday, 6 = Sunday)
df['dayofweek'] = df['Date_Time'].dt.dayofweek

# 3. Weekend flag (1 for Sat/Sun, 0 for Mon-Fri)
df['is_weekend'] = (df['dayofweek'] >= 5).astype(int)

# 4. Month (1 to 12) - needed for splitting the last month!
df['month'] = df['Date_Time'].dt.month
df
```

```python
# Training set: January to November (Months 1 to 11)
train_df = df[df['month'] < 12].copy()

# Test set: December (Month 12 - Last Month as required by assignment)
test_df = df[df['month'] == 12].copy()

print(f"Training samples (Jan - Nov): {len(train_df)}")
print(f"Testing samples (December): {len(test_df)}")
```

```python
df
```

```python
# List of input columns we want the model to learn from
features = [
    'Usage_kWh', 
    'Lagging_Current_Reactive.Power_kVarh',
    'Leading_Current_Reactive_Power_kVarh', 
    'CO2(tCO2)',
    'Lagging_Current_Power_Factor', 
    'Leading_Current_Power_Factor', 
    'hour', 
    'dayofweek', 
    'is_weekend'
]

# Separate features (X) and label (y)
X_train = train_df[features]
y_train = train_df['Load_Type']

X_test = test_df[features]
y_test = test_df['Load_Type']

print("Features X_train shape:", X_train.shape)
print("Target y_train shape:", y_train.shape)
```

```python
df
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, f1_score, classification_report

# 3 Models to Compare
models = {
    "Logistic Regression (Baseline)": Pipeline([
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(max_iter=2000, class_weight="balanced"))
    ]),
    "Random Forest": RandomForestClassifier(
        n_estimators=200, class_weight="balanced", 
        random_state=42, n_jobs=-1
    ),
    "HistGradientBoosting": HistGradientBoostingClassifier(
        class_weight="balanced", random_state=42
    ),
}

# Train all 3 and compare on December
rows = []
for name, m in models.items():
    m.fit(X_train, y_train)
    p = m.predict(X_test)
    rep = classification_report(y_test, p, output_dict=True)["macro avg"]
    rows.append({
        "Model": name,
        "Accuracy": round(accuracy_score(y_test, p), 4),
        "Macro Precision": round(rep["precision"], 4),
        "Macro Recall": round(rep["recall"], 4),
        "Macro F1": round(rep["f1-score"], 4)
    })

# Display as a clean comparison table
results_table = pd.DataFrame(rows).set_index("Model")
print(results_table)
```

```python
# Get the best model already trained
best_model = models["Random Forest"]
y_pred_final = best_model.predict(X_test)

print("=== FINAL MODEL: Random Forest ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred_final) * 100:.2f}%")
print(f"Macro F1: {f1_score(y_test, y_pred_final, average='macro'):.4f}\n")
print(classification_report(y_test, y_pred_final))
```

```python
df['Lagging_Current_Power_Factor'] = df['Lagging_Current_Power_Factor'].clip(0, 100)
df['Leading_Current_Power_Factor'] = df['Leading_Current_Power_Factor'].clip(0, 100)
df['NSM'] = df['NSM'].clip(0, 86400)


df['Usage_kWh_lag1'] = df['Usage_kWh'].shift(1).bfill()
df['Usage_kWh_rolling_mean_4'] = df['Usage_kWh'].rolling(window=4, min_periods=1).mean()

# Total Apparent Power
reactive_total = df['Lagging_Current_Reactive.Power_kVarh'] + df['Leading_Current_Reactive_Power_kVarh']
df['Apparent_Power'] = np.sqrt(df['Usage_kWh']**2 + reactive_total**2)


train_df = df[df['month'] < 12].copy()
test_df  = df[df['month'] == 12].copy()


features = [
    'Usage_kWh', 
    'Lagging_Current_Reactive.Power_kVarh',
    'Leading_Current_Reactive_Power_kVarh', 
    'CO2(tCO2)',
    'Lagging_Current_Power_Factor', 
    'Leading_Current_Power_Factor', 
    'NSM',
    'hour', 
    'dayofweek', 
    'is_weekend',
    'Usage_kWh_lag1',
    'Usage_kWh_rolling_mean_4',
    'Apparent_Power'
]

X_train = train_df[features]
y_train = train_df['Load_Type']

X_test = test_df[features]
y_test = test_df['Load_Type']


model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)


y_pred = model.predict(X_test)

acc = accuracy_score(y_test, y_pred)
print(f"New Overall Accuracy on December Test Set: {acc * 100:.2f}%\n")
print("--- New Classification Report ---")
print(classification_report(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

fig, ax = plt.subplots(figsize=(7, 6))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_final,
    display_labels=['Light_Load', 'Maximum_Load', 'Medium_Load'],
    cmap='Blues', ax=ax
)
plt.title('Random Forest — Confusion Matrix (December Test Set)')
plt.tight_layout()
plt.show()
```

```python
# Feature importance from Random Forest
rf_model = models["Random Forest"]
importance = pd.Series(
    rf_model.feature_importances_, 
    index=X_train.columns
).sort_values(ascending=True)

plt.figure(figsize=(9, 6))
importance.plot(kind='barh', color='steelblue')
plt.title('Feature Importance — Random Forest')
plt.xlabel('Importance Score')
plt.tight_layout()
plt.show()
```
